4 - Given three number x, y, and z,


       Output: x to the power of z,
               x to the power of y,
               x minus y,
               remainder value when x is divided by z

```python
x = int(input("Enter x: "))
y = int(input("Enter y: "))
z = int(input("Enter z: "))
print("x to the power of z: {}".format(x**z))
print("x to the power of y: {}".format(x**y))
print("x minus y: {}".format(x-y))
print("remainer value when x is divided by z".format(x%z))
```

###Comparision Operators

15 - Compare two number a =150, b = 225 and print if they are:

        equal,
        greater,
        greater than or equal to,
        less,
        less than or equal to,  
        not equal
       
        example : a < b
                  print("a is less than b")

```python
a = 150
b = 225
if a == b:
  print("a is equal to b")
else:
  print("a and b are not equal")

if a > b:
  print("a is greater than b")
  print("b is less than a")
elif b > a:
  print("b is greater than a")
  print("a is less than b")
else:
  print("a and b are equal")

if a >= b:
  print ("a is greater than or equal to b")
  print("b is less than or equal to a")
elif b >= a:
  print("b is greater than or equal to b")
  print("a is less than or equal to a")
```
